#### 9.10.4.1.1. Carga inicial y visión general

El archivo pesa aproximadamente **1.6 GB** y tiene **151 columnas**. Si se carga completo en pandas, las columnas de texto ocupan varias veces el tamaño del archivo en memoria, más de lo que tiene disponible un computador de 8 GB de RAM. Por eso el CSV se lee **por bloques** (`chunksize=50_000`) con `pd.read_csv`. Los bloques son pequeños para que el pico de memoria al leer cada bloque sea bajo:

- **Todas** las filas del archivo pasan por el ciclo; no se descarta ni se muestrea ninguna.
- En cada bloque se cuentan los estados del préstamo y los valores nulos de **las 151 columnas**, para tener la visión general del archivo completo.
- De cada bloque se conservan los préstamos finalizados (`Fully Paid` / `Charged Off`) con las columnas que se analizarán en el EDA.

**Columnas del EDA.** Se eligen las variables que se conocen **en el momento de otorgar el préstamo**. Las variables que se generan después de su emisión (`total_pymnt`, `recoveries`, `last_pymnt_d`, `last_fico_range_high`, `out_prncp`, `debt_settlement_flag`, `hardship_*`, `settlement_*`, ...) describen lo que pasó con el préstamo. Si se incluyeran, el modelo "vería el futuro" (*data leakage*), así que se excluyen. Tampoco se analizan identificadores o texto libre (`url`, `desc`, `emp_title`, `title`, `zip_code`).

````python
ruta = 'accepted_2007_to_2018Q4.csv'

num_cols_csv = ['loan_amnt', 'int_rate', 'installment', 'annual_inc', 'dti',
                'fico_range_low', 'fico_range_high', 'delinq_2yrs', 'inq_last_6mths',
                'open_acc', 'pub_rec', 'revol_bal', 'revol_util', 'total_acc',
                'mort_acc', 'pub_rec_bankruptcies', 'mths_since_last_delinq']
cat_cols = ['term', 'grade', 'sub_grade', 'home_ownership', 'verification_status',
            'purpose', 'addr_state', 'application_type', 'initial_list_status']
otras_cols = ['id', 'loan_status', 'issue_d', 'emp_length', 'earliest_cr_line']
columnas_eda = otras_cols + num_cols_csv + cat_cols
````

````python
t0 = time.time()
n_filas = 0
bloques = []
conteo_status = pd.Series(dtype=float)
nulos_fin = None                        # nulos por columna (151) en préstamos finalizados
nulos_fin_default = None                # nulos por columna solo en préstamos con default
nulos_anio, filas_anio = [], []         # nulos por año de emisión
nulos_por_fila = pd.Series(dtype=float) # cuántas filas tienen k nulos

for chunk in pd.read_csv(ruta, chunksize=50_000, low_memory=False):
    if n_filas == 0:
        primeras_filas = chunk.head()
        tipos = chunk.dtypes
    ultimas_filas = chunk.tail()
    n_filas += len(chunk)
    conteo_status = conteo_status.add(chunk['loan_status'].fillna('(vacío)').value_counts(), fill_value=0)

    fin = chunk[chunk['loan_status'].isin(['Fully Paid', 'Charged Off'])]
    nulos = fin.isna()
    es_default = fin['loan_status'] == 'Charged Off'
    nulos_fin = nulos.sum() if nulos_fin is None else nulos_fin + nulos.sum()
    nulos_fin_default = nulos[es_default].sum() if nulos_fin_default is None else nulos_fin_default + nulos[es_default].sum()
    anio = pd.to_datetime(fin['issue_d'], format='%b-%Y').dt.year
    nulos_anio.append(nulos.groupby(anio).sum())
    filas_anio.append(nulos.groupby(anio).size())
    nulos_por_fila = nulos_por_fila.add(nulos.sum(axis=1).value_counts(), fill_value=0)

    bloques.append(fin[columnas_eda])

df = pd.concat(bloques, ignore_index=True)
del bloques, chunk, fin, nulos

nulos_anio = pd.concat(nulos_anio).groupby(level=0).sum()
filas_anio = pd.concat(filas_anio).groupby(level=0).sum()

t_carga_pandas = time.time() - t0
print(f'Tiempo de lectura (pandas, por bloques): {t_carga_pandas:.1f} s')
print(f'Archivo completo:      {n_filas:,} filas x {len(tipos)} columnas')
print(f'Préstamos finalizados: {df.shape[0]:,} filas x {df.shape[1]} columnas seleccionadas')
````

#### 9.10.4.1.2. Análisis unidimensional (por variable)

##### Variables numéricas

Para cada variable numérica se calculan media, mediana, desviación estándar, mínimo, máximo, percentiles 25/50/75 y rango intercuartil (IQR). Los *outliers* se identifican con la regla del IQR: un valor es atípico si queda fuera del intervalo $[Q_1 - 1.5\,IQR,\; Q_3 + 1.5\,IQR]$. La tabla incluye también la asimetría, la curtosis y el porcentaje de faltantes.

````python
def resumen_numerico(s):
    q1, q2, q3 = s.quantile([0.25, 0.50, 0.75])
    iqr = q3 - q1
    lim_inf, lim_sup = q1 - 1.5 * iqr, q3 + 1.5 * iqr
    n_out = ((s < lim_inf) | (s > lim_sup)).sum()
    return pd.Series({'media': s.mean(), 'mediana (P50)': q2, 'desv_est': s.std(),
                      'min': s.min(), 'P25 (Q1)': q1, 'P75 (Q3)': q3, 'max': s.max(),
                      'IQR': iqr, 'lim_inf': lim_inf, 'lim_sup': lim_sup,
                      'n_outliers': n_out, '% outliers': 100 * n_out / s.notna().sum(),
                      'asimetria': s.skew(), 'curtosis': s.kurt(),
                      '% faltantes': 100 * s.isna().mean()})

tabla_num = pd.DataFrame({c: resumen_numerico(df[c]) for c in num_cols}).T
tabla_num.round(2)
````

- **Outliers (regla IQR):**
  - Las variables de conteo de eventos negativos (`delinq_2yrs` 19.3 %, `pub_rec` 16.9 %, `pub_rec_bankruptcies` 12.5 %) tienen muchos "outliers", pero son un artefacto: su Q1 = Q3 = 0, el IQR vale 0 y **cualquier valor distinto de 0 queda marcado como atípico**. **Normalidad.** La prueba de Shapiro-Wilk de `scipy` solo da p-valores confiables hasta $n \approx 5000$. Por eso se usa la prueba de **D'Agostino-Pearson** (`stats.normaltest`), que combina asimetría y curtosis y funciona con muestras grandes. Además se sigue una regla práctica: si $|\text{asimetría}| > 1$, la distribución es muy asimétrica y conviene considerar una transformación (log o Box-Cox).

````python
filas = []
for c in num_cols:
    s = df[c].dropna()
    k2, p = stats.normaltest(s)
    asim = s.skew()
    if abs(asim) > 1 and s.min() >= 0:
        sugerencia = 'log(1+x) / Box-Cox'
    elif abs(asim) > 1:
        sugerencia = 'Yeo-Johnson (tiene valores <= 0)'
    else:
        sugerencia = 'no necesaria'
    filas.append({'variable': c, 'asimetria': asim, 'curtosis': s.kurt(),
                  'estadistico K2': k2, 'p-value': p, 'transformacion sugerida': sugerencia})
tabla_normalidad = pd.DataFrame(filas).set_index('variable').sort_values('asimetria', ascending=False)
tabla_normalidad.round(3)
````

Efecto de la transformación `log(1+x)` sobre las variables más asimétricas:

````python
asimetricas = tabla_normalidad[tabla_normalidad['transformacion sugerida'] == 'log(1+x) / Box-Cox'].index[:8]
fig, axes = plt.subplots(2, len(asimetricas), figsize=(3.2 * len(asimetricas), 6))
for j, c in enumerate(asimetricas):
    s = df[c].dropna()
    axes[0, j].hist(s, bins=50, color='steelblue')
    axes[0, j].set_title(f'{c}\nasim={s.skew():.1f}')
    axes[1, j].hist(np.log1p(s), bins=50, color='darkorange')
    axes[1, j].set_title(f'log1p({c})\nasim={np.log1p(s).skew():.1f}')
plt.tight_layout()
plt.show()
````

**Comentario sobre normalidad y transformaciones.**

- La prueba de D'Agostino-Pearson rechaza la normalidad en **todas** las variables (p ≈ 0). Con 1.3 millones de datos era de esperarse: la prueba detecta hasta la mínima desviación, así que se usa la asimetría como criterio práctico.
- **Transformación recomendada (log(1+x) o Box-Cox):** `annual_inc`, `revol_bal`, `installment`, `mort_acc`, `open_acc`, `inq_last_6mths` y `credit_hist_years`. La figura lo confirma: la asimetría de `annual_inc` pasa de 46 a valores cercanos a 0 con `log1p`, y `revol_bal` queda casi simétrica.
- `dti` tiene valores ≤ 0 (−1), así que `log` no aplica directamente. Primero hay que corregir los valores inválidos (−1 y 999) y luego aplicar `log1p` o Yeo-Johnson.
- En `pub_rec`, `delinq_2yrs` y `pub_rec_bankruptcies` el log no arregla nada, porque la mayoría de los valores son 0. Se recomienda **binarizarlas** (`tiene_registro = valor > 0`).
- Para `fico_range_*` la asimetría (1.29) se debe al límite inferior de la política de crédito. Basta con estandarizar (`StandardScaler`).
- Estas transformaciones importan sobre todo para los modelos lineales (`LogisticRegression`, `LinearSVC`) y para `NaiveBayes` gaussiano. Los árboles (`DecisionTree`, `RandomForest`, `GBT`) son invariantes a transformaciones monótonas.

**Pruebas de diferencia entre clases.** Para cada variable numérica se calculan:

- La diferencia de medias entre `default = 1` y `default = 0`.
- La prueba **t de Student** en su versión de Welch (`equal_var=False`), que no supone varianzas iguales.
- La prueba de **Mann-Whitney U**, que no supone normalidad. Es la adecuada aquí, porque ninguna variable es normal.
- La correlación **punto-biserial** $r_{pb}$ entre la variable y `default`.

Con $n > 1.3$ millones casi cualquier diferencia sale "estadísticamente significativa". Por eso, para juzgar la **relevancia práctica** se usan la magnitud de $r_{pb}$ y la diferencia de medianas, y no solo el p-value.

````python
filas = []
for c in num_cols:
    datos = df[[c, 'default']].dropna()
    g0 = datos.loc[datos['default'] == 0, c]
    g1 = datos.loc[datos['default'] == 1, c]
    t, p_t = stats.ttest_ind(g1, g0, equal_var=False)
    u, p_u = stats.mannwhitneyu(g1, g0, alternative='two-sided')
    r, p_r = stats.pointbiserialr(datos['default'], datos[c])
    filas.append({'variable': c, 'media_0': g0.mean(), 'media_1': g1.mean(),
                  'dif_medias': g1.mean() - g0.mean(),
                  'mediana_0': g0.median(), 'mediana_1': g1.median(),
                  't (Welch)': t, 'p t': p_t, 'p Mann-Whitney': p_u,
                  'r punto-biserial': r, 'p r': p_r,
                  'significativa (p<0.05)': 'sí' if p_u < 0.05 else 'no'})
tabla_pruebas = pd.DataFrame(filas).set_index('variable')
tabla_pruebas = tabla_pruebas.reindex(tabla_pruebas['r punto-biserial'].abs().sort_values(ascending=False).index)
tabla_pruebas.round(4)
````

````python
r_pb = tabla_pruebas['r punto-biserial'].sort_values()
plt.figure(figsize=(8, 7))
plt.barh(r_pb.index, r_pb.values, color=np.where(r_pb.values > 0, 'indianred', 'steelblue'))
plt.axvline(0, color='black', lw=0.8)
plt.title('Correlación punto-biserial con default')
plt.xlabel('r_pb')
plt.show()
````

**Comentario de las pruebas.**

- **Todas** las diferencias son estadísticamente significativas (p ≈ 0 en t de Welch y en Mann-Whitney), incluso en variables cuyas medianas son idénticas (`emp_length`, `total_acc`). Con n = 1.3 M el p-value deja de discriminar: hasta una diferencia de 0.1 años sale significativa.
- Por eso la **relevancia** se juzga con $r_{pb}$:
  - **Mayor asociación:** `int_rate` (r = **0.259**), `fico_range_low/high` (r = **−0.131**), `dti` (0.085), `mort_acc` (−0.075), `loan_amnt` (0.066), `inq_last_6mths` (0.066) y `revol_util` (0.060).
  - **Asociación casi nula (|r| < 0.03):** `total_acc`, `mths_since_last_delinq`, `emp_length`, `delinq_2yrs`, `revol_bal`, `pub_rec`, `pub_rec_bankruptcies` y `open_acc`.
- Todas las correlaciones son **débiles** (|r| < 0.3). Ninguna variable predice el default por sí sola, así que los modelos tendrán que combinar muchas variables y posiblemente relaciones no lineales. Es razonable esperar un AUC moderado, no cercano a 1.
- **Signos:** a mayor tasa, deuda relativa, monto y consultas de crédito recientes, mayor riesgo. A mayor FICO, ingreso, número de hipotecas y antigüedad crediticia, menor riesgo. Todos los signos son coherentes con la intuición financiera.

**Prueba chi-cuadrado de independencia** entre cada variable categórica y `default`. Como con $n$ tan grande el p-value será casi siempre ≈ 0, se reporta también la **V de Cramér**, que mide la fuerza de la asociación en una escala de 0 a 1:

$$V = \sqrt{\frac{\chi^2 / n}{\min(k-1,\, r-1)}}$$

````python
def v_cramer(x, y):
    ct = pd.crosstab(x, y)
    chi2, p, dof, _ = stats.chi2_contingency(ct)
    n = ct.values.sum()
    return chi2, p, dof, np.sqrt((chi2 / n) / (min(ct.shape) - 1))

filas = []
for c in cat_cols:
    chi2, p, dof, v = v_cramer(df[c], df['default'])
    filas.append({'variable': c, 'chi2': chi2, 'gl': dof, 'p-value': p, 'V de Cramer': v,
                  'independiente (p>=0.05)': 'sí' if p >= 0.05 else 'no'})
tabla_chi2 = pd.DataFrame(filas).set_index('variable').sort_values('V de Cramer', ascending=False)
tabla_chi2.round(4)
````

**Redundancia entre `grade` / `sub_grade` e `int_rate`.** Lending Club asigna la tasa de interés según el grado del préstamo. Para medir qué parte de la variación de `int_rate` explica `grade` se usa la razón de correlación $\eta^2$ (varianza entre grupos / varianza total):

````python
def eta_cuadrado(cat, y):
    media_grupo = y.groupby(cat, observed=True).transform('mean')
    return ((media_grupo - y.mean()) ** 2).sum() / ((y - y.mean()) ** 2).sum()

print(f"eta^2 int_rate ~ grade:     {eta_cuadrado(df['grade'], df['int_rate']):.3f}")
print(f"eta^2 int_rate ~ sub_grade: {eta_cuadrado(df['sub_grade'], df['int_rate']):.3f}")

plt.figure(figsize=(10, 4.5))
sns.boxplot(data=df, x='grade', y='int_rate', color='steelblue', showfliers=False)
plt.title('int_rate por grade')
plt.show()
````

**Relación de los faltantes con `default` (¿MCAR?).** Si los faltantes fueran *completamente al azar* (MCAR), la tasa de default debería ser la misma en las filas donde la variable falta y en las filas donde no falta. Para cada columna con nulos se construye la tabla 2×2 (nulo / no nulo × default) y se aplica la prueba chi-cuadrado:

- $H_0$: el hecho de que falte el dato es independiente de `default`, lo que es consistente con MCAR.
- p < 0.05: se rechaza $H_0$. El faltante **no** es MCAR y el hecho de que falte ya es informativo.

Se calcula para todas las columnas con nulos del archivo, usando los conteos acumulados en la lectura.

````python
n_def = df['default'].sum()
filas = []
for c in con_nulos.index:
    if nulos_fin[c] == n_fin:          # 100 % nulos: no hay nada que comparar
        continue
    nulo_def = nulos_fin_default[c]
    nulo_nodef = nulos_fin[c] - nulo_def
    tabla = np.array([[nulo_nodef, nulo_def],
                      [(n_fin - n_def) - nulo_nodef, n_def - nulo_def]])
    chi2, p, _, _ = stats.chi2_contingency(tabla)
    filas.append({'variable': c, '% nulos': 100 * nulos_fin[c] / n_fin,
                  'tasa_default si nulo %': 100 * nulo_def / nulos_fin[c],
                  'tasa_default si no nulo %': 100 * (n_def - nulo_def) / (n_fin - nulos_fin[c]),
                  'chi2': chi2, 'p-value': p})
tabla_mcar = pd.DataFrame(filas).set_index('variable')
tabla_mcar['dif (pp)'] = tabla_mcar['tasa_default si nulo %'] - tabla_mcar['tasa_default si no nulo %']
tabla_mcar['compatible con MCAR'] = np.where(tabla_mcar['p-value'] >= 0.05, 'sí', 'no')
print(tabla_mcar['compatible con MCAR'].value_counts())
with pd.option_context('display.max_rows', 200):
    display(tabla_mcar.round(4))
````

Detalle para las variables del EDA que tienen nulos:

````python
eda_con_nulos = [c for c in num_cols + cat_cols if df[c].isna().any()]
filas = []
for c in eda_con_nulos:
    falta = df[c].isna()
    ct = pd.crosstab(falta, df['default'])
    chi2, p, _, _ = stats.chi2_contingency(ct)
    filas.append({'variable': c, 'n nulos': falta.sum(), '% nulos': 100 * falta.mean(),
                  'tasa_default si nulo %': 100 * df.loc[falta, 'default'].mean(),
                  'tasa_default si no nulo %': 100 * df.loc[~falta, 'default'].mean(),
                  'p-value chi2': p})
tabla_mcar_eda = pd.DataFrame(filas).set_index('variable').sort_values('% nulos', ascending=False)
display(tabla_mcar_eda.round(4))

print('% de nulos por año de emisión (variables del EDA con nulos):')
display((100 * df[eda_con_nulos].isna().groupby(df['issue_year']).mean()).round(2).T)
````

#### 9.10.4.2.1. Partición común de los datos

La prueba de DeLong compara AUC calculados sobre **las mismas observaciones**. Por eso scikit-learn y PySpark deben usar exactamente la misma partición de entrenamiento y prueba. La partición se hace así:

1. Se define **una sola vez**: 80 % entrenamiento / 20 % prueba, **estratificada** por `default` y con semilla fija `random_state=42`.
2. Se conserva el identificador único de cada préstamo (`id`) y se guarda la asignación (`id`, `split`) en el archivo `resultados/particion_train_test.csv`.
3. Tanto scikit-learn (este cuaderno) como PySpark **leen ese mismo archivo** para separar los conjuntos.

El archivo se guarda en CSV y no en Parquet, porque escribir Parquet con pandas requiere instalar una librería adicional (`pyarrow`). PySpark lee CSV directamente (`spark.read.csv(..., header=True)`), así que el resultado es el mismo.

````python
import os
import gc
from sklearn.model_selection import train_test_split

os.makedirs('resultados', exist_ok=True)
t0_prep_sklearn = time.time()   # inicio del preprocesamiento (scikit-learn)

df['id'] = pd.to_numeric(df['id']).astype('int64')
print('¿id es único?', df['id'].is_unique)

idx_train, idx_test = train_test_split(df.index, test_size=0.2, stratify=df['default'],
                                       random_state=42)

particion = pd.DataFrame({'id': df['id'], 'split': 'train'})
particion.loc[idx_test, 'split'] = 'test'
particion.to_csv('resultados/particion_train_test.csv', index=False)
print('Partición guardada en resultados/particion_train_test.csv')
````

Ahora se **lee la asignación desde el disco**, como lo hará PySpark, y se une a `df` por `id`. Así se garantiza que el modelado usa exactamente lo que quedó guardado en el archivo.

````python
particion = pd.read_csv('resultados/particion_train_test.csv')
df = df.merge(particion, on='id', how='left', validate='one_to_one')

es_train = df['split'] == 'train'
es_test = df['split'] == 'test'

resumen_particion = df.groupby('split')['default'].agg(n='size', n_default='sum', tasa_default='mean')
resumen_particion['%_filas'] = 100 * resumen_particion['n'] / len(df)
resumen_particion['tasa_default'] = (100 * resumen_particion['tasa_default']).round(3)
resumen_particion
````

- `open_acc`: r = 0.70 con `total_acc`.
- `grade` y `sub_grade`: explican más del 90 % de la varianza de `int_rate`. Se conserva `int_rate`, que es numérica y tiene la mayor correlación con `default`.

- `application_type` e `initial_list_status`: asociación prácticamente nula con `default`.

**Categorías raras.**

- En `home_ownership`, `ANY` y `NONE` se unen a `OTHER`.
- En `purpose` y `addr_state`, las categorías con menos del 1 % de los préstamos **del conjunto de entrenamiento** se agrupan en `other` / `OTROS`.

**Regla principal:** todo lo que *aprende* de los datos se ajusta **solo con el conjunto de entrenamiento** y después se aplica al de prueba:

- la lista de categorías raras;
- las medianas del `SimpleImputer`;
- las medias y desviaciones del `StandardScaler`;
- las categorías del `OneHotEncoder`.

Las transformaciones fila a fila (`log1p`, indicadores, recortes con valores fijos) no aprenden nada de los datos, así que se pueden aplicar a todo el DataFrame.

````python
X = pd.DataFrame(index=df.index)

for c in ['loan_amnt', 'int_rate', 'fico_range_high', 'emp_length', 'inq_last_6mths',
          'mort_acc', 'total_acc', 'credit_hist_years']:
    X[c] = df[c]
X['dti'] = df['dti'].where(df['dti'].between(0, 100))      # fuera de [0, 100] -> NaN
X['revol_util'] = df['revol_util'].clip(upper=100)
X['log_annual_inc'] = np.log1p(df['annual_inc'])
X['log_revol_bal'] = np.log1p(df['revol_bal'])
X['tiene_delinq'] = (df['delinq_2yrs'] > 0).astype(float)
X['tiene_pub_rec'] = (df['pub_rec'] > 0).astype(float)
X['nunca_moroso'] = df['mths_since_last_delinq'].isna().astype(float)
X['emp_length_faltante'] = df['emp_length'].isna().astype(float)
vars_num = list(X.columns)

X['term'] = df['term'].astype(str)
X['verification_status'] = df['verification_status'].astype(str)
X['home_ownership'] = df['home_ownership'].astype(str).replace({'ANY': 'OTHER', 'NONE': 'OTHER'})
for c, nombre_otros in [('purpose', 'other'), ('addr_state', 'OTROS')]:
    frec_train = df.loc[es_train, c].value_counts(normalize=True)   # solo entrenamiento
    raras = list(frec_train[frec_train < 0.01].index)
    X[c] = df[c].astype(str).where(~df[c].isin(raras), nombre_otros)
    print(f'{c}: {len(raras)} categorías raras agrupadas en "{nombre_otros}" -> {raras}')
vars_cat = ['term', 'verification_status', 'home_ownership', 'purpose', 'addr_state']

print(f'\n{len(vars_num)} variables numéricas: {vars_num}')
print(f'{len(vars_cat)} variables categóricas: {vars_cat}')
````

**Imputación, escalado y codificación.**

- Las variables numéricas se imputan con la **mediana** y luego se estandarizan con `StandardScaler`, que es necesario para `LogisticRegression` y `LinearSVC`.
- Las categóricas se codifican con `OneHotEncoder` (`handle_unknown='ignore'`: si en prueba apareciera una categoría no vista, simplemente queda con todas las columnas en 0).

Se usa una única matriz escalada para los seis modelos. Los árboles son invariantes a transformaciones monótonas como el escalado, así que el resultado es el mismo y se evita tener dos copias de los datos en memoria. Además, la matriz se guarda en `float32`, que ocupa la mitad de memoria que `float64`.

````python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

X_train_df, X_test_df = X.loc[es_train], X.loc[es_test]
y_train = df.loc[es_train, 'default'].values
y_test = df.loc[es_test, 'default'].values
id_test = df.loc[es_test, 'id'].values

imputador = SimpleImputer(strategy='median').fit(X_train_df[vars_num])        # solo train
escalador = StandardScaler().fit(imputador.transform(X_train_df[vars_num]))   # solo train
codificador = OneHotEncoder(handle_unknown='ignore', sparse_output=False,
                            dtype=np.float32).fit(X_train_df[vars_cat])        # solo train

def transformar(datos):
    num = escalador.transform(imputador.transform(datos[vars_num])).astype(np.float32)
    cat = codificador.transform(datos[vars_cat])
    return np.hstack([num, cat])

X_train = transformar(X_train_df)
X_test = transformar(X_test_df)
nombres_variables = vars_num + list(codificador.get_feature_names_out(vars_cat))

del X, X_train_df, X_test_df
gc.collect()
t_prep_sklearn = time.time() - t0_prep_sklearn
print(f'Tiempo de preprocesamiento (partición + variables + transformadores): {t_prep_sklearn:.1f} s')

print('X_train:', X_train.shape, f'({X_train.nbytes / 2**20:.0f} MB)')
print('X_test: ', X_test.shape, f'({X_test.nbytes / 2**20:.0f} MB)')
print('Tasa de default  train: %.4f | test: %.4f' % (y_train.mean(), y_test.mean()))
print('Medianas usadas en la imputación:')
print(pd.Series(imputador.statistics_, index=vars_num).round(3).to_string())
````

### 9.10.4.3. Modelos y espacios de búsqueda

Se implementan los seis modelos con los espacios de búsqueda del enunciado y el mismo criterio de selección (AUC ROC) que en PySpark:

| Modelo | scikit-learn | PySpark | Espacio de búsqueda |
|---|---|---|---|
| Regresión logística | `LogisticRegression` | `LogisticRegression` | `regParam` ∈ {1e-6, 1e-5, 1e-4}; en scikit-learn $C = 1/(\text{regParam} \times n)$, con $n$ = tamaño del entrenamiento |
| Árbol de decisión | `DecisionTreeClassifier` | `DecisionTreeClassifier` | `max_depth` ∈ {5, 10, 15} |
| Bosque aleatorio | `RandomForestClassifier` | `RandomForestClassifier` | `n_estimators` ∈ {10, 50, 100} y `max_depth` ∈ {5, 10, 15} |
| Gradient boosting | `GradientBoostingClassifier` | `GBTClassifier` | `n_estimators` ∈ {50, 100} y `max_depth` ∈ {3, 5}, `learning_rate` = 0.1 |
| SVM lineal | `LinearSVC(loss="hinge")` | `LinearSVC` | los mismos valores de $C$ que la regresión logística |
| Naive Bayes | `GaussianNB` | `NaiveBayes(modelType="gaussian")` | sin búsqueda (configuración por defecto) |

**Consideraciones de equivalencia (lado scikit-learn):**

- **Regularización L2** en la regresión logística y en LinearSVC (`penalty='l2'`, que es el valor por defecto).
- **$C = 1/(\text{regParam} \times n)$.** La relación es **aproximada**: en scikit-learn, $C$ multiplica la *suma* de las pérdidas, mientras que en PySpark `regParam` multiplica el término de regularización frente al *promedio* de las pérdidas. Además, dentro de la validación cruzada cada modelo se entrena con 2/3 de $n$, así que la equivalencia exacta solo se da en el ajuste final.
- **`LinearSVC(loss="hinge")`:** el valor por defecto de scikit-learn es `squared_hinge`, pero PySpark minimiza la pérdida hinge, así que se fija `loss="hinge"`. Con esta pérdida liblinear exige `dual=True`.
- **Árboles:** scikit-learn evalúa **todos los cortes exactos** de cada variable continua, mientras que PySpark discretiza cada variable en `maxBins` intervalos (32 por defecto). Con pocos intervalos, los árboles de PySpark pueden encontrar cortes algo peores, pero entrenan más rápido.
- **Naive Bayes gaussiano** en ambos entornos, porque las variables estandarizadas tienen valores negativos que el NB multinomial o de Bernoulli no admiten.
- **Gradient boosting:** se usa el `GradientBoostingClassifier` exacto que pide la tabla (no `HistGradientBoostingClassifier`).
- **Sin pesos de clase**, para que ambos entornos entrenen exactamente el mismo problema. El desbalance (20 % de default) se maneja con la estratificación y con el AUC como criterio de selección, que no depende del umbral.

````python
from sklearn.linear_model import LogisticRegression
from sklearn.svm import LinearSVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.naive_bayes import GaussianNB

n = X_train.shape[0]
reg_params = [1e-6, 1e-5, 1e-4]
valores_C = [1 / (r * n) for r in reg_params]
print(f'n = {n:,}')
for r, c in zip(reg_params, valores_C):
    print(f'regParam = {r:.0e}  ->  C = {c:.6f}')

modelos = {
    'LogisticRegression': (LogisticRegression(penalty='l2', max_iter=1000),
                           {'C': valores_C}),
    'DecisionTree': (DecisionTreeClassifier(random_state=42),
                     {'max_depth': [5, 10, 15]}),
    'RandomForest': (RandomForestClassifier(random_state=42),
                     {'n_estimators': [10, 50, 100], 'max_depth': [5, 10, 15]}),
    'GradientBoosting': (GradientBoostingClassifier(learning_rate=0.1, random_state=42),
                         {'n_estimators': [50, 100], 'max_depth': [3, 5]}),
    'LinearSVC': (LinearSVC(loss='hinge', penalty='l2', dual=True, max_iter=1000, random_state=42),
                  {'C': valores_C}),
    'GaussianNB': (GaussianNB(), None),
}
````

**Procedimiento para cada modelo:**

1. `GridSearchCV` **sin `Pipeline`** (los datos ya están transformados), con `cv=3` (el mismo número de pliegues que en PySpark) y `scoring='roc_auc'`.

   **Sobre `n_jobs`:** el enunciado indica `n_jobs=-1` (usar los 12 núcleos lógicos), pero se usa `n_jobs=4`. Cada trabajador en paralelo hace su propia copia del pliegue de entrenamiento (~720 mil filas). Además, `LogisticRegression` (lbfgs) y `LinearSVC` (liblinear) convierten los datos a `float64`, así que cada copia ocupa entre 350 y 500 MB. Con 12 trabajadores eso supera los 7.7 GB de RAM del equipo. En la primera ejecución con `n_jobs=-1` se produjo un `MemoryError`. Con 4 trabajadores la búsqueda es la misma (mismos pliegues, mismos hiperparámetros, mismos resultados); solo tarda más.
2. Con el mejor modelo (reentrenado sobre todo el conjunto de entrenamiento, `refit=True`) se predice el conjunto de prueba.
3. Se calculan Accuracy, Precision, Recall, F1 y ROC AUC, más la matriz de confusión.
4. Se guarda la **puntuación continua** de cada observación de prueba (`predict_proba(X_test)[:, 1]`, o `decision_function(X_test)` en `LinearSVC`), que se usará en la prueba de DeLong.
5. Se mide con `time.time()` el tiempo de entrenamiento (búsqueda completa con validación cruzada + reentrenamiento final) y el de predicción.

Las clases predichas (`predict`) usan el umbral por defecto: probabilidad ≥ 0.5, o `decision_function` ≥ 0 en LinearSVC. El ROC AUC, en cambio, no depende del umbral.

GaussianNB no tiene hiperparámetros que buscar, así que se entrena directamente. Por coherencia, igual se reporta su AUC de validación cruzada con `cross_val_score` (cv=3).

````python
from sklearn.model_selection import GridSearchCV, cross_val_score
from sklearn.metrics import (accuracy_score, precision_score, recall_score, f1_score,
                             roc_auc_score, confusion_matrix)

resultados = {}
scores_test = pd.DataFrame({'id': id_test, 'default': y_test})

def entrenar_y_evaluar(nombre):
    modelo, grid = modelos[nombre]
    t0 = time.time()
    if grid is not None:
        busqueda = GridSearchCV(modelo, grid, cv=3, scoring='roc_auc', n_jobs=N_JOBS)
        busqueda.fit(X_train, y_train)
        mejor = busqueda.best_estimator_
        mejores_params, auc_cv = busqueda.best_params_, busqueda.best_score_
        tabla_cv = pd.DataFrame(busqueda.cv_results_)[['params', 'mean_test_score', 'std_test_score',
                                                        'mean_fit_time', 'rank_test_score']]
    else:
        auc_cv = cross_val_score(modelo, X_train, y_train, cv=3, scoring='roc_auc', n_jobs=N_JOBS).mean()
        mejor = modelo.fit(X_train, y_train)
        mejores_params, tabla_cv = {}, None
    t_entrenamiento = time.time() - t0

    t0 = time.time()
    y_pred = mejor.predict(X_test)
    if hasattr(mejor, 'predict_proba'):
        score = mejor.predict_proba(X_test)[:, 1]
    else:
        score = mejor.decision_function(X_test)
    t_prediccion = time.time() - t0

    scores_test[f'score_{nombre}'] = score
    resultados[nombre] = {
        'mejores_params': mejores_params,
        'AUC_cv': auc_cv,
        'Accuracy': accuracy_score(y_test, y_pred),
        'Precision': precision_score(y_test, y_pred, zero_division=0),
        'Recall': recall_score(y_test, y_pred),
        'F1': f1_score(y_test, y_pred),
        'ROC_AUC': roc_auc_score(y_test, score),
        't_entrenamiento_s': t_entrenamiento,
        't_prediccion_s': t_prediccion,
        'matriz_confusion': confusion_matrix(y_test, y_pred),
        'modelo': mejor,
    }
    r = resultados[nombre]
    print(f'{nombre}: mejores hiperparámetros = {mejores_params}')
    print(f'  AUC CV = {auc_cv:.4f} | AUC test = {r["ROC_AUC"]:.4f} | Accuracy = {r["Accuracy"]:.4f} | '
          f'Precision = {r["Precision"]:.4f} | Recall = {r["Recall"]:.4f} | F1 = {r["F1"]:.4f}')
    print(f'  Tiempo de entrenamiento = {t_entrenamiento:.1f} s | tiempo de predicción = {t_prediccion:.2f} s')
    if tabla_cv is not None:
        display(tabla_cv.sort_values('rank_test_score').round(4))
````

**Diagnóstico de LinearSVC.** El AUC de prueba de LinearSVC (0.47) es peor que el azar, y el modelo clasifica todos los préstamos como `0`. No es un error de código ni de convergencia: es una propiedad de la **pérdida hinge** en este problema.

- La pérdida hinge es $\max(0,\, 1 - y_i f(x_i))$ con $y_i \in \{-1, +1\}$.
- La solución trivial $w = 0$, $b = -1$ deja **sin pérdida al 80 % de las observaciones** (todos los pagados) y cuesta 2 por cada préstamo en default.
- Como las clases se solapan mucho (en el EDA ninguna variable tiene $|r| > 0.26$), mover $w$ en cualquier dirección aumenta la pérdida de muchos negativos más de lo que la reduce en los positivos. Por eso el óptimo **colapsa a $w \approx 0$** y las puntuaciones quedan prácticamente constantes. El pequeño $w$ que queda es ruido numérico, así que el orden de las puntuaciones, y con él el AUC, es arbitrario.

La celda siguiente lo comprueba con tres evidencias:

1. Las puntuaciones de prueba son casi constantes.
2. El valor de la función objetivo en la solución de liblinear es prácticamente igual al de la solución trivial.
3. Si se da más peso a la clase minoritaria (`class_weight='balanced'`), el mismo LinearSVC hinge sí aprende (AUC ≈ 0.71, igual que la regresión logística).

El modelo que se reporta y se compara es el **sin pesos**, porque es el que pide el enunciado y el que se entrena en PySpark. El modelo balanceado aparece aquí solo como referencia y no se incluye en las comparaciones.

````python
svc = resultados['LinearSVC']['modelo']
ys = np.where(y_train == 1, 1, -1)      # la pérdida hinge usa etiquetas -1 / +1

def objetivo_hinge(w, b, C):
    # 0.5 ||w||^2 + C * suma de pérdidas hinge (liblinear además regulariza b, efecto despreciable)
    return 0.5 * (w @ w) + C * np.maximum(0, 1 - ys * (X_train @ w + b)).sum()

score_svc = scores_test['score_LinearSVC']
print(f'C elegido: {svc.C:.6f}')
print(f'Puntuaciones en test: min = {score_svc.min():.4f}, max = {score_svc.max():.4f}, '
      f'desviación estándar = {score_svc.std():.6f}')
print(f'Objetivo hinge en la solución de liblinear:           {objetivo_hinge(svc.coef_[0], svc.intercept_[0], svc.C):,.2f}')
print(f'Objetivo hinge en la solución trivial (w = 0, b = -1): {objetivo_hinge(np.zeros(X_train.shape[1]), -1.0, svc.C):,.2f}')

t0 = time.time()
control = LinearSVC(loss='hinge', C=svc.C, class_weight='balanced', dual=True,
                    max_iter=1000, random_state=42).fit(X_train, y_train)
auc_control = roc_auc_score(y_test, control.decision_function(X_test))
print(f'\nReferencia: LinearSVC hinge con class_weight="balanced" -> AUC test = {auc_control:.4f} '
      f'({time.time() - t0:.0f} s)')
````

**Sesión de Spark.** Se usa la configuración del enunciado con tres ajustes, justificados por el hardware y por problemas encontrados al ejecutar:

- `spark.driver.memory = 4g` en lugar de `8g`: el equipo tiene 7.7 GB de RAM en total. Reservar 8 GB para la JVM dejaría sin memoria al sistema operativo y a Python.
- Spark corre en **modo local** (`local[*]`, un solo equipo con 12 núcleos lógicos). En este modo el *driver* y el *executor* son el mismo proceso, así que `spark.executor.memory` no tiene efecto. Se fija en `4g` solo por coherencia.
- `spark.memory.fraction = 0.6` en lugar de `0.8`. Esa fracción del heap se reserva para caché y ejecución, y el resto queda para los objetos de los modelos. Los bosques aleatorios de Spark crean un objeto Java por cada nodo de cada árbol: en una prueba, un bosque de solo 10 árboles de profundidad 15 tuvo unos 168 mil nodos, y el grid llega a 100 árboles. Con 4 GB de heap y 0.8, a esos objetos les quedarían solo ~0.7 GB.
- `spark.local.dir` en el disco `D:`. Es la carpeta donde Spark escribe los datos que no caben en memoria (`MEMORY_AND_DISK`). En una ejecución anterior, los temporales de Spark (casi 7 GB, que no se borraron al interrumpirse las sesiones) **llenaron el disco `C:`**, y la sesión de Spark se cayó a mitad del entrenamiento. `D:` tiene cientos de GB libres. Si `D:` no existe, se usa la carpeta temporal del sistema.

El resto (`shuffle.partitions = 400`, `default.parallelism = 400`, `memory.storageFraction = 0.3`) se mantiene igual al enunciado.

Las categorías raras de `purpose` y `addr_state` (menos del 1 %) se calculan **solo con el conjunto de entrenamiento**. Para eso se trae al *driver* únicamente la lista de nombres de esas categorías, que es un resultado agregado de unas pocas filas y no el dataset.

````python
for c in ['loan_amnt', 'int_rate', 'fico_range_high', 'inq_last_6mths', 'mort_acc', 'total_acc',
          'dti', 'revol_util', 'annual_inc', 'revol_bal', 'delinq_2yrs', 'pub_rec', 'mths_since_last_delinq']:
    sdf = sdf.withColumn(c, F.col(c).cast('double'))

def a_fecha(c):
    return F.to_date(F.concat(F.lit('01-'), F.col(c)), 'dd-MMM-yyyy')

sdf = (sdf
    .withColumn('emp_length', F.when(F.col('emp_length') == '< 1 year', 0.0)
                               .otherwise(F.regexp_extract('emp_length', r'(\d+)', 1).cast('double')))
    .withColumn('credit_hist_years', F.datediff(a_fecha('issue_d'), a_fecha('earliest_cr_line')) / 365.25)
    .withColumn('dti', F.when(F.col('dti').between(0, 100), F.col('dti')))          # fuera de [0, 100] -> nulo
    .withColumn('revol_util', F.when(F.col('revol_util') > 100, 100.0).otherwise(F.col('revol_util')))
    .withColumn('log_annual_inc', F.log1p('annual_inc'))
    .withColumn('log_revol_bal', F.log1p('revol_bal'))
    .withColumn('tiene_delinq', F.when(F.col('delinq_2yrs') > 0, 1.0).otherwise(0.0))
    .withColumn('tiene_pub_rec', F.when(F.col('pub_rec') > 0, 1.0).otherwise(0.0))
    .withColumn('nunca_moroso', F.col('mths_since_last_delinq').isNull().cast('double'))
    .withColumn('emp_length_faltante', F.col('emp_length').isNull().cast('double'))
    .withColumn('term', F.trim('term'))
    .withColumn('home_ownership', F.when(F.col('home_ownership').isin('ANY', 'NONE'), 'OTHER')
                                   .otherwise(F.col('home_ownership'))))

vars_num_s = vars_num          # mismas 16 variables numéricas que en scikit-learn
vars_cat_s = vars_cat          # mismas 5 variables categóricas

n_train_s = sdf.filter(F.col('split') == 'train').count()
for c, nombre_otros in [('purpose', 'other'), ('addr_state', 'OTROS')]:
    frec = (sdf.filter(F.col('split') == 'train').groupBy(c).count()
               .filter(F.col('count') / n_train_s < 0.01))
    raras = sorted(r[c] for r in frec.select(c).collect())      # solo la lista de categorías raras
    sdf = sdf.withColumn(c, F.when(F.col(c).isin(raras), nombre_otros).otherwise(F.col(c)))
    print(f'{c}: {len(raras)} categorías raras agrupadas en "{nombre_otros}" -> {raras}')
````

**Reparticionado.** Tras la lectura `multiLine` y el *join* (que Spark resuelve con *broadcast* porque la tabla de partición es pequeña), todos los datos quedan en **una sola partición**, es decir, en un solo núcleo. Por eso se reparticionan explícitamente.

Se usan **48 particiones** (4 por núcleo lógico) en lugar de las 400 de `spark.default.parallelism`. En una prueba previa con los mismos datos, entrenar con 400 particiones fue de 3 a 6 veces más lento: por ejemplo, LogisticRegression tardó 29 s contra 9 s, y un árbol de profundidad 10 tardó 62 s contra 12 s. Con ~1 millón de filas, 400 particiones dejan solo ~2 700 filas por tarea, y el costo de coordinar tantas tareas en cada iteración supera el cálculo útil. La configuración `shuffle.partitions = 400` se mantiene para las operaciones de *shuffle* (*joins*, agregaciones).

**Transformadores (ajustados solo con train)** en un `Pipeline` de Spark:

1. `Imputer` (mediana) para las variables numéricas.
2. `StringIndexer` (`handleInvalid='keep'`) + `OneHotEncoder` (`dropLast=True`) para las categóricas. Con esta combinación, el índice extra de "categoría no vista" es justo la columna que se descarta. Así, las categorías reales conservan todas su columna (igual que en scikit-learn) y una categoría desconocida queda con todo en 0, equivalente a `handle_unknown='ignore'`. El vector final tiene las mismas 62 columnas.
3. `VectorAssembler` de las numéricas → `StandardScaler` (`withMean=True`, `withStd=True`), aplicado solo a las numéricas, como en scikit-learn.
4. `VectorAssembler` final (numéricas escaladas + dummies) → columna `features`.

El resultado se **cachea obligatoriamente** con `.persist(StorageLevel.MEMORY_AND_DISK)` antes de entrenar, y se materializa con `count()`.

````python
from pyspark.ml import Pipeline
from pyspark.ml.feature import (Imputer, StringIndexer, OneHotEncoder, VectorAssembler,
                                StandardScaler as StandardScalerSpark)
from pyspark.ml.feature import StandardScalerModel, StringIndexerModel, ImputerModel

N_PARTICIONES = 48
sdf = sdf.select(['id', 'default', 'split'] + vars_num_s + vars_cat_s).repartition(N_PARTICIONES)
train_s = sdf.filter(F.col('split') == 'train')
test_s = sdf.filter(F.col('split') == 'test')

etapas = ([Imputer(inputCols=vars_num_s, outputCols=[c + '_imp' for c in vars_num_s], strategy='median')]
          + [StringIndexer(inputCol=c, outputCol=c + '_idx', handleInvalid='keep') for c in vars_cat_s]
          + [OneHotEncoder(inputCols=[c + '_idx' for c in vars_cat_s],
                           outputCols=[c + '_ohe' for c in vars_cat_s], dropLast=True),
             VectorAssembler(inputCols=[c + '_imp' for c in vars_num_s], outputCol='num_vec'),
             StandardScalerSpark(inputCol='num_vec', outputCol='num_esc', withMean=True, withStd=True),
             VectorAssembler(inputCols=['num_esc'] + [c + '_ohe' for c in vars_cat_s], outputCol='features')])

prep_spark = Pipeline(stages=etapas).fit(train_s)                  # solo train
train_final = prep_spark.transform(train_s).select('id', 'default', 'features')
test_final = prep_spark.transform(test_s).select('id', 'default', 'features')

train_final = train_final.persist(StorageLevel.MEMORY_AND_DISK)   # caché obligatorio
test_final = test_final.persist(StorageLevel.MEMORY_AND_DISK)
n_train_final, n_test_final = train_final.count(), test_final.count()
t_prep_spark = time.time() - t0_prep_spark

escalador_spark = next(s for s in prep_spark.stages if isinstance(s, StandardScalerModel))
indexadores_spark = [s for s in prep_spark.stages if isinstance(s, StringIndexerModel)]
imputador_spark = next(s for s in prep_spark.stages if isinstance(s, ImputerModel))
nombres_variables_spark = vars_num_s + [f'{c}_{etiqueta}' for c, ind in zip(vars_cat_s, indexadores_spark)
                                         for etiqueta in ind.labelsArray[0]]

print(f'train: {n_train_final:,} filas | test: {n_test_final:,} filas | particiones: {train_final.rdd.getNumPartitions()}')
print(f'Dimensión del vector de variables: {len(nombres_variables_spark)}')
print(f'Tiempo de preprocesamiento en Spark (lectura + variables + transformadores + caché): {t_prep_spark:.1f} s')
print('Medianas usadas en la imputación (Spark):')
print(imputador_spark.surrogateDF.toPandas().T.round(3).rename(columns={0: 'mediana'}).to_string())
````

### 9.10.4.5. Modelado con PySpark

Se entrenan los seis modelos de `pyspark.ml.classification` con los mismos espacios de búsqueda de 9.10.4.3. La búsqueda se hace **exclusivamente con `ParamGridBuilder` + `CrossValidator`** (`numFolds=3`, igual que en scikit-learn), sin bucles sobre combinaciones de hiperparámetros.

**Detalles de implementación:**

- **Criterio de selección:** `BinaryClassificationEvaluator(metricName='areaUnderROC')`.
  - Para los modelos que entregan probabilidades, el evaluador usa la columna **`probability`**. En los árboles de Spark, `rawPrediction` contiene los **conteos** de cada clase en la hoja y no probabilidades. En una prueba previa, evaluar el árbol con `rawPrediction` daba AUC ≈ 0.49 (orden sin sentido) y con `probability` ≈ 0.70.
  - Para LinearSVC, que no produce probabilidades, se usa `rawPrediction` (el margen $w^\top x + b$).
- **Regresión logística:** `elasticNetParam=0` (L2 pura) y `regParam` ∈ {1e-6, 1e-5, 1e-4}.
- **LinearSVC:** minimiza la pérdida hinge con los mismos `regParam`.
- **Árboles** (DecisionTree, RandomForest, GBT): `maxBins=32` (valor por defecto). Cada variable continua se discretiza en a lo sumo 32 intervalos antes de buscar los cortes. scikit-learn, en cambio, evalúa todos los cortes posibles.
- **RandomForest: `maxMemoryInMB=32` y dos combinaciones excluidas por capacidad de cómputo insuficiente.**
  - *Memoria.* En una primera ejecución, el bosque aleatorio agotó el heap de la JVM (`java.lang.OutOfMemoryError: Java heap space`) y la sesión de Spark se cerró. En modo local se ejecutan 12 tareas a la vez. Cada tarea recibe una copia de los árboles que se están construyendo y reserva hasta `maxMemoryInMB` (256 MB por defecto) para los histogramas de los nodos, así que 12 tareas pueden pedir más de 3 GB de un heap de 4 GB. Con `maxMemoryInMB=32`, cada pasada sobre los datos procesa menos nodos a la vez (más pasadas), sin cambiar los hiperparámetros del modelo. Con ese ajuste, un bosque de 10 árboles de profundidad 15 se entrenó sin problemas, con ~1 GB de heap.
  - *Tiempo.* En este equipo, cada árbol de profundidad 15 le tomó a Spark ~60 s con todos los datos: 10 árboles, unos 168 mil nodos, tardaron 601 s. Las combinaciones **50 × 15** y **100 × 15** suman 150 árboles de profundidad 15 en cada uno de los 3 pliegues, más el reentrenamiento final. Solo ellas tomarían **alrededor de 6 horas**, y el grid completo de Spark entre 9 y 10 horas. scikit-learn entrenó el grid completo de RandomForest en ~15 minutos. Por eso el grid de Spark usa **7 de las 9 combinaciones**: `numTrees` ∈ {10, 50, 100} con `maxDepth` ∈ {5, 10}, más 10 × 15. La exclusión se hace uniendo dos listas generadas con `ParamGridBuilder`, sin bucles manuales.
  - Esta restricción se debe a la **capacidad de cómputo insuficiente** de un solo equipo con 7.7 GB de RAM en modo local, y se retoma en las conclusiones. En un clúster, cada nodo aportaría su propia memoria y núcleos. La comparación de RandomForest entre entornos debe leerse con esta diferencia en mente: el espacio de búsqueda de Spark es más pequeño.
- **GBT:** `stepSize=0.1`, igual al `learning_rate` de scikit-learn.
- **Naive Bayes:** `modelType='gaussian'`, sin hiperparámetros. Se usa un `ParamGridBuilder().build()` vacío, así que el `CrossValidator` solo calcula el AUC de validación cruzada.
- **Semilla:** `seed=42` en todos los modelos aleatorios y en el `CrossValidator`. Los pliegues de Spark se generan con su propia semilla, así que no son los mismos 3 pliegues de scikit-learn. La partición train/test sí es idéntica.

**Métricas y tiempos:**

- Se calculan las mismas métricas que en scikit-learn. La matriz de confusión se construye con un `groupBy('default', 'prediction').count()`, que devuelve solo 4 filas agregadas.
- Una vez terminados el entrenamiento y la predicción, se transfieren al *driver* **únicamente** las columnas `id`, `default` y la puntuación de la clase positiva del conjunto de prueba, extraída con `vector_to_array`. El tiempo de esa transferencia se mide por separado.

````python
import pyspark.ml.classification as clf_spark
from pyspark.ml.tuning import ParamGridBuilder, CrossValidator
from pyspark.ml.evaluation import BinaryClassificationEvaluator
from pyspark.ml.functions import vector_to_array

evaluador_prob = BinaryClassificationEvaluator(labelCol='default', rawPredictionCol='probability',
                                               metricName='areaUnderROC')
evaluador_raw = BinaryClassificationEvaluator(labelCol='default', rawPredictionCol='rawPrediction',
                                              metricName='areaUnderROC')

lr_s = clf_spark.LogisticRegression(labelCol='default', featuresCol='features', elasticNetParam=0.0)
dt_s = clf_spark.DecisionTreeClassifier(labelCol='default', featuresCol='features', seed=42)
rf_s = clf_spark.RandomForestClassifier(labelCol='default', featuresCol='features', seed=42,
                                        maxMemoryInMB=32)   # ver explicación arriba
gbt_s = clf_spark.GBTClassifier(labelCol='default', featuresCol='features', stepSize=0.1, seed=42)
svc_s = clf_spark.LinearSVC(labelCol='default', featuresCol='features')
nb_s = clf_spark.NaiveBayes(labelCol='default', featuresCol='features', modelType='gaussian')

modelos_spark = {
    'LogisticRegression': (lr_s, ParamGridBuilder().addGrid(lr_s.regParam, reg_params).build(), evaluador_prob),
    'DecisionTree': (dt_s, ParamGridBuilder().addGrid(dt_s.maxDepth, [5, 10, 15]).build(), evaluador_prob),
    # 50 x 15 y 100 x 15 se excluyen por capacidad de cómputo insuficiente (ver explicación arriba)
    'RandomForest': (rf_s, ParamGridBuilder().addGrid(rf_s.numTrees, [10, 50, 100])
                                             .addGrid(rf_s.maxDepth, [5, 10]).build()
                           + ParamGridBuilder().addGrid(rf_s.numTrees, [10])
                                               .addGrid(rf_s.maxDepth, [15]).build(), evaluador_prob),
    'GradientBoosting': (gbt_s, ParamGridBuilder().addGrid(gbt_s.maxIter, [50, 100])
                                                  .addGrid(gbt_s.maxDepth, [3, 5]).build(), evaluador_prob),
    'LinearSVC': (svc_s, ParamGridBuilder().addGrid(svc_s.regParam, reg_params).build(), evaluador_raw),
    'GaussianNB': (nb_s, ParamGridBuilder().build(), evaluador_prob),
}
print('maxBins usado en los árboles de Spark:', dt_s.getMaxBins(), rf_s.getMaxBins(), gbt_s.getMaxBins())
for nombre, (_, grid, _) in modelos_spark.items():
    print(f'{nombre}: {len(grid)} combinaciones x 3 pliegues')
````

````python
resultados_spark = {}
scores_test_spark = pd.DataFrame({'id': id_test, 'default': y_test})

def entrenar_y_evaluar_spark(nombre):
    estimador, grid, evaluador = modelos_spark[nombre]
    col_score = 'rawPrediction' if nombre == 'LinearSVC' else 'probability'
    cv = CrossValidator(estimator=estimador, estimatorParamMaps=grid, evaluator=evaluador,
                        numFolds=3, seed=42)
    t0 = time.time()
    cv_modelo = cv.fit(train_final)
    t_entrenamiento = time.time() - t0
    mejor = cv_modelo.bestModel
    params = [{p.name: v for p, v in g.items()} for g in grid]
    tabla_cv = pd.DataFrame({'params': params, 'AUC_cv': cv_modelo.avgMetrics})
    i_mejor = int(np.argmax(cv_modelo.avgMetrics))

    t0 = time.time()
    pred = (mejor.transform(test_final)
                 .withColumn('score', vector_to_array(col_score)[1])
                 .select('id', 'default', 'prediction', col_score, 'score')
                 .persist(StorageLevel.MEMORY_AND_DISK))
    auc = evaluador.evaluate(pred)
    conteos = pred.groupBy('default', 'prediction').count().toPandas()      # 4 filas agregadas
    t_prediccion = time.time() - t0

    t0 = time.time()
    local = pred.select('id', 'default', 'score').toPandas()               # solo id, default, score de test
    t_transferencia = time.time() - t0
    pred.unpersist()

    n = {(int(r['default']), int(r['prediction'])): int(r['count']) for _, r in conteos.iterrows()}
    vn, fp, fn, vp = n.get((0, 0), 0), n.get((0, 1), 0), n.get((1, 0), 0), n.get((1, 1), 0)
    precision = vp / (vp + fp) if vp + fp > 0 else 0.0
    recall = vp / (vp + fn)
    scores_test_spark[f'score_{nombre}'] = scores_test_spark['id'].map(local.set_index('id')['score'])

    resultados_spark[nombre] = {
        'mejores_params': params[i_mejor], 'AUC_cv': cv_modelo.avgMetrics[i_mejor],
        'Accuracy': (vp + vn) / (vp + vn + fp + fn), 'Precision': precision, 'Recall': recall,
        'F1': 2 * precision * recall / (precision + recall) if precision + recall > 0 else 0.0,
        'ROC_AUC': auc, 't_entrenamiento_s': t_entrenamiento, 't_prediccion_s': t_prediccion,
        't_transferencia_s': t_transferencia,
        'matriz_confusion': np.array([[vn, fp], [fn, vp]]), 'modelo': mejor}
    r = resultados_spark[nombre]
    print(f'{nombre}: mejores hiperparámetros = {params[i_mejor]}')
    print(f'  AUC CV = {r["AUC_cv"]:.4f} | AUC test = {auc:.4f} | Accuracy = {r["Accuracy"]:.4f} | '
          f'Precision = {precision:.4f} | Recall = {recall:.4f} | F1 = {r["F1"]:.4f}')
    print(f'  Tiempo de entrenamiento (CV + ajuste final) = {t_entrenamiento:.1f} s | '
          f'predicción = {t_prediccion:.2f} s | transferencia al driver = {t_transferencia:.2f} s')
    display(tabla_cv.sort_values('AUC_cv', ascending=False).round(4))
````

### 9.10.4.6. Comparación estadística de los modelos

#### 9.10.4.6.1. Prueba de DeLong

La prueba de DeLong (DeLong, DeLong y Clarke-Pearson, 1988) contrasta $H_0: \text{AUC}_1 = \text{AUC}_2$ cuando ambos AUC se calculan **sobre las mismas observaciones**. Como las dos curvas ROC están correlacionadas, el error estándar de la diferencia incluye su covarianza:

$$z = \frac{\widehat{AUC}_1 - \widehat{AUC}_2}{\sqrt{\operatorname{Var}(\widehat{AUC}_1) + \operatorname{Var}(\widehat{AUC}_2) - 2\operatorname{Cov}(\widehat{AUC}_1, \widehat{AUC}_2)}}$$

**Implementación.** Se programa la **versión rápida de Sun y Xu (2014)**, de complejidad $O(n \log n)$. El AUC y las componentes estructurales de DeLong se obtienen a partir de **rangos medios** (`scipy.stats.rankdata`):

- los rangos de las puntuaciones de los positivos entre sí ($T_X$);
- los de los negativos entre sí ($T_Y$);
- los de todas las puntuaciones juntas ($T_Z$).

Con ellos, $V_{10} = (T_Z^{pos} - T_X)/n$ y $V_{01} = 1 - (T_Z^{neg} - T_Y)/m$, y la covarianza de los AUC es $S_{10}/m + S_{01}/n$. La versión directa, que compara cada positivo con cada negativo, es $O(m \cdot n)$: con ~54 mil positivos y ~215 mil negativos serían unos 11 600 millones de comparaciones por modelo, lo que no es viable.

La implementación se **valida primero con un conjunto pequeño**, comparándola con la versión directa.

````python
def delong_rapido(y, scores):
    """DeLong rápido (Sun y Xu, 2014), O(n log n).
    y: etiquetas 0/1 (n,). scores: matriz (k, n) con las puntuaciones de k modelos.
    Devuelve los k AUC y su matriz de covarianza (k x k)."""
    scores = np.atleast_2d(scores)
    pos, neg = y == 1, y == 0
    m, n = int(pos.sum()), int(neg.sum())       # enteros de Python: evitan desbordamiento en m * n
    k = scores.shape[0]
    tx, ty, tz = np.empty((k, m)), np.empty((k, n)), np.empty((k, m + n))
    for r in range(k):
        x, yy = scores[r, pos], scores[r, neg]
        tx[r] = stats.rankdata(x)                         # rangos medios entre positivos
        ty[r] = stats.rankdata(yy)                        # rangos medios entre negativos
        tz[r] = stats.rankdata(np.concatenate([x, yy]))   # rangos medios en el total
    aucs = (tz[:, :m].sum(axis=1) - m * (m + 1) / 2) / (m * n)
    v01 = (tz[:, :m] - tx) / n                            # componente estructural de cada positivo
    v10 = 1 - (tz[:, m:] - ty) / m                        # componente estructural de cada negativo
    return aucs, np.atleast_2d(np.cov(v01)) / m + np.atleast_2d(np.cov(v10)) / n


def delong_directo(y, scores):
    """Versión directa O(m*n) (DeLong et al., 1988). Solo para validar con pocos datos."""
    scores = np.atleast_2d(scores)
    pos, neg = y == 1, y == 0
    m, n = int(pos.sum()), int(neg.sum())
    k = scores.shape[0]
    v10, v01, aucs = np.empty((k, m)), np.empty((k, n)), np.empty(k)
    for r in range(k):
        x, yy = scores[r, pos][:, None], scores[r, neg][None, :]
        psi = (x > yy) + 0.5 * (x == yy)                  # matriz m x n de comparaciones
        aucs[r] = psi.mean()
        v10[r], v01[r] = psi.mean(axis=1), psi.mean(axis=0)
    return aucs, np.atleast_2d(np.cov(v10)) / m + np.atleast_2d(np.cov(v01)) / n


def prueba_delong(y, score_1, score_2):
    """Prueba de DeLong para dos AUC calculados sobre las mismas observaciones."""
    aucs, cov = delong_rapido(y, np.vstack([score_1, score_2]))
    dif = aucs[0] - aucs[1]
    ee_dif = np.sqrt(cov[0, 0] + cov[1, 1] - 2 * cov[0, 1])
    ee = np.sqrt(np.diag(cov))
    z = dif / ee_dif
    return {'AUC_1': aucs[0], 'IC95_1': (aucs[0] - 1.96 * ee[0], aucs[0] + 1.96 * ee[0]),
            'AUC_2': aucs[1], 'IC95_2': (aucs[1] - 1.96 * ee[1], aucs[1] + 1.96 * ee[1]),
            'dif_AUC': dif, 'IC95_dif': (dif - 1.96 * ee_dif, dif + 1.96 * ee_dif),
            'z': z, 'p': 2 * stats.norm.sf(abs(z))}


def holm(p_valores):
    """Corrección de Holm para comparaciones múltiples."""
    p = np.asarray(p_valores, dtype=float)
    ajustados, maximo = np.empty(len(p)), 0.0
    for posicion, i in enumerate(np.argsort(p)):
        maximo = max(maximo, min(1.0, (len(p) - posicion) * p[i]))
        ajustados[i] = maximo
    return ajustados
````

**Validación** con 3 000 observaciones de prueba elegidas al azar (semilla 42) y tres modelos: la versión rápida debe coincidir con la directa.

````python
rng = np.random.default_rng(42)
idx_val = rng.choice(len(y_test), 3000, replace=False)
S_val = np.vstack([scores_test[f'score_{m}'].values[idx_val]
                   for m in ['GradientBoosting', 'LogisticRegression', 'DecisionTree']])
auc_r, cov_r = delong_rapido(y_test[idx_val], S_val)
auc_d, cov_d = delong_directo(y_test[idx_val], S_val)
print('AUC (rápida): ', auc_r.round(8))
print('AUC (directa):', auc_d.round(8))
print(f'Máxima diferencia en AUC:        {np.abs(auc_r - auc_d).max():.2e}')
print(f'Máxima diferencia en covarianza: {np.abs(cov_r - cov_d).max():.2e}')
print('p-value GB vs LR  (rápida / directa):',
      round(prueba_delong(y_test[idx_val], S_val[0], S_val[1])['p'], 6),
      '/', round(2 * stats.norm.sf(abs((auc_d[0] - auc_d[1]) /
                                       np.sqrt(cov_d[0, 0] + cov_d[1, 1] - 2 * cov_d[0, 1]))), 6))
````

**Margen de relevancia práctica (definido de antemano).** Con ~269 mil observaciones de prueba, el error estándar de una diferencia de AUC es del orden de 0.001, así que diferencias muy pequeñas pueden resultar estadísticamente significativas sin importar en la práctica. Se fija un margen de **|ΔAUC| ≥ 0.005**: una diferencia menor que medio punto de AUC no cambia de forma apreciable el orden de riesgo de los solicitantes ni las decisiones de crédito. Una diferencia se considera **relevante** solo si es estadísticamente significativa (p ajustado por Holm < 0.05) **y** supera ese margen.

Se construye un vector común de etiquetas uniendo las puntuaciones de ambos entornos por `id`.

````python
MARGEN_RELEVANCIA = 0.005
ALFA = 0.05
nombres_modelos = list(modelos.keys())

comun = scores_test.merge(scores_test_spark, on='id', suffixes=('_sk', '_sp'))
assert (comun['default_sk'] == comun['default_sp']).all() and len(comun) == len(y_test)
y_comun = comun['default_sk'].values
score_sk = {m: comun[f'score_{m}_sk'].values for m in nombres_modelos}
score_sp = {m: comun[f'score_{m}_sp'].values for m in nombres_modelos}
print(f'Observaciones comunes: {len(comun):,} | positivos: {y_comun.sum():,}')

def tabla_delong(pares, familia):
    filas = []
    for etiqueta, s1, s2 in pares:
        r = prueba_delong(y_comun, s1, s2)
        filas.append({'comparación': etiqueta,
                      'AUC_1': r['AUC_1'], 'IC95_1': f"[{r['IC95_1'][0]:.4f}, {r['IC95_1'][1]:.4f}]",
                      'AUC_2': r['AUC_2'], 'IC95_2': f"[{r['IC95_2'][0]:.4f}, {r['IC95_2'][1]:.4f}]",
                      'dif_AUC': r['dif_AUC'], 'IC95_dif': f"[{r['IC95_dif'][0]:.4f}, {r['IC95_dif'][1]:.4f}]",
                      'z': r['z'], 'p': r['p']})
    t = pd.DataFrame(filas).set_index('comparación')
    t['p_Holm'] = holm(t['p'])
    t['significativa'] = np.where(t['p_Holm'] < ALFA, 'sí', 'no')
    t['relevante'] = np.where((t['p_Holm'] < ALFA) & (t['dif_AUC'].abs() >= MARGEN_RELEVANCIA), 'sí', 'no')
    t.insert(0, 'familia', familia)
    return t
````

**Mapas de calor por entorno.** A la izquierda, la diferencia de AUC (fila − columna) con el AUC de cada modelo en la diagonal. A la derecha, los p-values ajustados por Holm.

````python
def matrices_delong(tabla, aucs):
    dif = pd.DataFrame(np.nan, index=nombres_modelos, columns=nombres_modelos)
    p_aj = dif.copy()
    for etiqueta, fila in tabla.iterrows():
        a, b = etiqueta.split(' vs ')
        dif.loc[a, b], dif.loc[b, a] = fila['dif_AUC'], -fila['dif_AUC']
        p_aj.loc[a, b] = p_aj.loc[b, a] = fila['p_Holm']
    for m in nombres_modelos:
        dif.loc[m, m] = aucs[m]
    return dif, p_aj

aucs_sk = {m: resultados[m]['ROC_AUC'] for m in nombres_modelos}
aucs_sp = {m: resultados_spark[m]['ROC_AUC'] for m in nombres_modelos}
fig, axes = plt.subplots(2, 2, figsize=(20, 16))
for fila, (entorno, tabla, aucs) in enumerate([('scikit-learn', delong_sk, aucs_sk),
                                               ('PySpark', delong_sp, aucs_sp)]):
    dif, p_aj = matrices_delong(tabla, aucs)
    anot = dif.map(lambda v: f'{v:.4f}') if hasattr(dif, 'map') else dif.applymap(lambda v: f'{v:.4f}')
    fuera_diag = dif.where(~np.eye(len(dif), dtype=bool))
    lim = np.nanmax(np.abs(fuera_diag.values))
    sns.heatmap(fuera_diag, annot=anot, fmt='', cmap='coolwarm', center=0, vmin=-lim, vmax=lim,
                ax=axes[fila, 0], cbar_kws={'label': 'ΔAUC (fila - columna)'})
    axes[fila, 0].set_title(f'{entorno}: ΔAUC entre modelos (diagonal = AUC)')
    sns.heatmap(p_aj, annot=True, fmt='.3g', cmap='RdYlGn_r', vmin=0, vmax=0.1, ax=axes[fila, 1],
                cbar_kws={'label': 'p ajustado (Holm)'})
    axes[fila, 1].set_title(f'{entorno}: p-values de DeLong ajustados por Holm')
plt.tight_layout()
plt.show()
````

#### 9.10.4.6.2. Pruebas complementarias: McNemar y bootstrap pareado

**Umbral de clasificación.** McNemar y el F1 comparan **decisiones**, así que necesitan un umbral. El umbral se elige **sin usar el conjunto de prueba**: para cada modelo se toma el que **maximiza el F1 en el conjunto de entrenamiento**, con el mismo procedimiento en ambos entornos:

1. Se generan 199 umbrales candidatos: los cuantiles 0.5 %, 1 %, …, 99.5 % de las puntuaciones de entrenamiento del modelo.
2. Para cada candidato se cuentan predichos positivos y verdaderos positivos, y se calcula el F1.
3. Se elige el umbral con mayor F1.

Se usa F1 porque es la métrica que equilibra precision y recall de la clase minoritaria (default). El umbral 0.5 deja un recall de apenas 4-8 % (ver 9.10.4.4).

En PySpark los cuantiles se obtienen con `approxQuantile` (error relativo 0.001) y los conteos con una sola agregación, sin traer al *driver* las puntuaciones de entrenamiento. Una limitación: las puntuaciones de entrenamiento son *dentro de muestra*, así que en modelos que se sobreajustan (por ejemplo, un bosque con árboles de profundidad 15) el umbral puede quedar algo optimista.

````python
NIVELES_UMBRAL = np.linspace(0.005, 0.995, 199)

def umbral_f1_sklearn(y, s):
    candidatos = np.unique(np.quantile(s, NIVELES_UMBRAL))
    positivos = y.sum()
    mejor = (0.0, candidatos[0])
    for t in candidatos:
        pred = s >= t
        vp = np.sum(pred & (y == 1))
        f1 = 2 * vp / (pred.sum() + positivos)
        mejor = max(mejor, (f1, t))
    return mejor[1], mejor[0]

def umbral_f1_spark(nombre):
    col_score = 'rawPrediction' if nombre == 'LinearSVC' else 'probability'
    s = (resultados_spark[nombre]['modelo'].transform(train_final)
         .select('default', vector_to_array(col_score)[1].alias('score'))
         .persist(StorageLevel.MEMORY_AND_DISK))
    candidatos = sorted(set(s.approxQuantile('score', list(NIVELES_UMBRAL), 0.001)))
    expresiones = [F.sum('default').alias('positivos')]
    for i, t in enumerate(candidatos):
        expresiones += [F.sum(F.when(F.col('score') >= t, 1).otherwise(0)).alias(f'pp_{i}'),
                        F.sum(F.when((F.col('score') >= t) & (F.col('default') == 1), 1).otherwise(0)).alias(f'vp_{i}')]
    fila = s.agg(*expresiones).first()          # una sola fila de conteos agregados
    s.unpersist()
    f1s = [2 * fila[f'vp_{i}'] / (fila[f'pp_{i}'] + fila['positivos']) for i in range(len(candidatos))]
    i = int(np.argmax(f1s))
    return candidatos[i], f1s[i]

t0 = time.time()
filas = []
umbral_sk, umbral_sp = {}, {}
for m in nombres_modelos:
    umbral_sk[m], f1_sk = umbral_f1_sklearn(y_train, scores_train_sklearn[f'score_{m}'].values)
    umbral_sp[m], f1_sp = umbral_f1_spark(m)
    filas.append({'modelo': m, 'umbral_sklearn': umbral_sk[m], 'F1_train_sklearn': f1_sk,
                  'umbral_pyspark': umbral_sp[m], 'F1_train_pyspark': f1_sp})
tabla_umbrales = pd.DataFrame(filas).set_index('modelo')
print(f'Tiempo: {time.time() - t0:.1f} s')
tabla_umbrales.round(4)
````

**Comparaciones incluidas** en ambas pruebas:

- (i) Las 6 comparaciones **entre entornos**: el mismo modelo en scikit-learn y en PySpark.
- (ii) En cada entorno, el par formado por los **dos modelos con mayor AUC**.

Se corrige con Holm dentro de cada familia, con α = 0.05.

**Prueba de McNemar.** Para cada par se cuentan las observaciones discordantes:

- $b$: acierta el modelo 1 y falla el modelo 2;
- $c$: falla el modelo 1 y acierta el modelo 2.

El estadístico es $\chi^2 = (|b - c| - 1)^2 / (b + c)$, con 1 grado de libertad y corrección de continuidad. Si $b + c < 25$ se usa la versión exacta binomial (`scipy.stats.binomtest`).

````python
def mcnemar(y, pred_1, pred_2):
    acierta_1, acierta_2 = pred_1 == y, pred_2 == y
    b = int(np.sum(acierta_1 & ~acierta_2))
    c = int(np.sum(~acierta_1 & acierta_2))
    if b + c == 0:
        return b, c, np.nan, 1.0, 'sin discordantes'
    if b + c < 25:
        return b, c, np.nan, stats.binomtest(min(b, c), b + c, 0.5).pvalue, 'exacta (binomial)'
    chi2 = (abs(b - c) - 1) ** 2 / (b + c)
    return b, c, chi2, stats.chi2.sf(chi2, df=1), 'chi2 con corrección'

top2_sk = list(tabla_resultados.index[:2])
top2_sp = list(tabla_resultados_spark.index[:2])
comparaciones = ([('entre entornos', f'{m}: sklearn vs PySpark', ('sk', m), ('sp', m)) for m in nombres_modelos]
                 + [('top-2 dentro del entorno', f'sklearn: {top2_sk[0]} vs {top2_sk[1]}', ('sk', top2_sk[0]), ('sk', top2_sk[1])),
                    ('top-2 dentro del entorno', f'PySpark: {top2_sp[0]} vs {top2_sp[1]}', ('sp', top2_sp[0]), ('sp', top2_sp[1]))])

def puntuacion(clave):
    entorno, m = clave
    return (score_sk if entorno == 'sk' else score_sp)[m]

def umbral(clave):
    entorno, m = clave
    return (umbral_sk if entorno == 'sk' else umbral_sp)[m]

filas = []
for familia, etiqueta, k1, k2 in comparaciones:
    pred_1 = (puntuacion(k1) >= umbral(k1)).astype(int)
    pred_2 = (puntuacion(k2) >= umbral(k2)).astype(int)
    b, c, chi2, p, metodo = mcnemar(y_comun, pred_1, pred_2)
    filas.append({'familia': familia, 'comparación': etiqueta, 'acc_1': np.mean(pred_1 == y_comun),
                  'acc_2': np.mean(pred_2 == y_comun), 'b': b, 'c': c, 'chi2': chi2, 'p': p, 'método': metodo})
tabla_mcnemar = pd.DataFrame(filas).set_index('comparación')
tabla_mcnemar['p_Holm'] = tabla_mcnemar.groupby('familia')['p'].transform(lambda p: holm(p.values))
tabla_mcnemar['significativa'] = np.where(tabla_mcnemar['p_Holm'] < ALFA, 'sí', 'no')
tabla_mcnemar.round(5)
````

**Bootstrap pareado.** Se remuestrea con reemplazo el conjunto de prueba **B = 2000** veces (semilla 42), usando **los mismos índices para todos los modelos**. En cada remuestreo se calculan AUC, AUC-PR (área bajo la curva precision-recall) y F1 (con el umbral de entrenamiento de cada modelo). Luego, para cada comparación se obtiene la diferencia media y su intervalo percentil del 95 %. Hay evidencia de diferencia cuando el intervalo **no contiene el cero**.

Implementación eficiente: un remuestreo con reemplazo equivale a dar a cada observación un **peso** igual al número de veces que fue elegida. Las puntuaciones de cada modelo se ordenan **una sola vez**, y en cada remuestreo solo cambian los pesos. Así, el AUC y el AUC-PR ponderados se calculan en $O(n)$ con sumas acumuladas, en lugar de reordenar 269 mil valores 2 000 veces. Antes de usarla se verifica que, con todos los pesos iguales a 1, la implementación coincide con `roc_auc_score`, `average_precision_score` y `f1_score` de scikit-learn.

````python
from sklearn.metrics import average_precision_score

class MetricasPonderadas:
    """AUC, AUC-PR y F1 con pesos de remuestreo, en O(n) por remuestreo."""
    def __init__(self, y, score, umbral):
        self.orden = np.argsort(-score, kind='mergesort')
        s = score[self.orden]
        self.y = y[self.orden].astype(float)
        self.grupos = np.r_[0, np.flatnonzero(np.diff(s) != 0) + 1]   # inicio de cada grupo de empates
        self.pred = (s >= umbral).astype(float)

    def calcular(self, w):
        w = w[self.orden]
        wp, wn = w * self.y, w * (1 - self.y)
        P, N = np.add.reduceat(wp, self.grupos), np.add.reduceat(wn, self.grupos)
        P_tot, N_tot = P.sum(), N.sum()
        auc = np.sum(P * (N_tot - np.cumsum(N) + 0.5 * N)) / (P_tot * N_tot)
        vp, fp = np.cumsum(P), np.cumsum(N)
        precision = np.divide(vp, vp + fp, out=np.zeros_like(vp), where=(vp + fp) > 0)
        auc_pr = np.sum(np.diff(np.r_[0.0, vp / P_tot]) * precision)
        f1 = 2 * np.sum(wp * self.pred) / (np.sum(w * self.pred) + P_tot)
        return auc, auc_pr, f1

# verificación con pesos = 1 frente a scikit-learn
m_ref = 'GradientBoosting'
calc = MetricasPonderadas(y_comun, score_sk[m_ref], umbral_sk[m_ref])
propio = calc.calcular(np.ones(len(y_comun)))
referencia = (roc_auc_score(y_comun, score_sk[m_ref]), average_precision_score(y_comun, score_sk[m_ref]),
              f1_score(y_comun, score_sk[m_ref] >= umbral_sk[m_ref]))
print('Implementación propia:', np.round(propio, 10))
print('scikit-learn:          ', np.round(referencia, 10))
````

````python
claves = sorted({k for _, _, k1, k2 in comparaciones for k in (k1, k2)})
calculadoras = {k: MetricasPonderadas(y_comun, puntuacion(k), umbral(k)) for k in claves}
boot = {k: np.empty((B_BOOTSTRAP, 3)) for k in claves}
rng = np.random.default_rng(42)
n_test = len(y_comun)
t0 = time.time()
for b in range(B_BOOTSTRAP):
    w = np.bincount(rng.integers(0, n_test, n_test), minlength=n_test).astype(float)  # mismos índices para todos
    for k, calc in calculadoras.items():
        boot[k][b] = calc.calcular(w)
print(f'{B_BOOTSTRAP} remuestreos x {len(claves)} modelos en {time.time() - t0:.0f} s')

filas = []
for familia, etiqueta, k1, k2 in comparaciones:
    dif = boot[k1] - boot[k2]
    fila = {'familia': familia, 'comparación': etiqueta}
    for j, metrica in enumerate(['AUC', 'AUC_PR', 'F1']):
        li, ls = np.percentile(dif[:, j], [2.5, 97.5])
        fila[f'dif_{metrica}'] = dif[:, j].mean()
        fila[f'IC95_{metrica}'] = f'[{li:.4f}, {ls:.4f}]'
        fila[f'evidencia_{metrica}'] = 'sí' if (li > 0 or ls < 0) else 'no'
    filas.append(fila)
tabla_bootstrap = pd.DataFrame(filas).set_index('comparación')
tabla_bootstrap.round(4)
````

### 9.10.4.7. Interpretabilidad con LIME

LIME (*Local Interpretable Model-agnostic Explanations*) explica **una predicción individual**:

1. Genera muchas versiones perturbadas de la observación (5 000).
2. Obtiene la predicción del modelo para cada una.
3. Ajusta un modelo lineal simple, ponderado por cercanía a la observación original.

Los coeficientes de ese modelo local indican qué variables empujaron la predicción hacia *default* o hacia *pagado*.

**Configuración:**

- Se explica el **modelo con mayor AUC de cada entorno entre los que dan probabilidades**. LinearSVC queda excluido porque no las produce.
- Las explicaciones se construyen en las **unidades originales** de las variables: tasa en %, FICO en puntos, etc. La función de predicción aplica internamente el mismo escalado que se usó al entrenar. Así, reglas como "int_rate > 15.99" son interpretables.
- Los indicadores binarios y las dummies se declaran como **categóricas**, para que LIME las perturbe solo con valores 0/1.
- Se eligen **dos préstamos de prueba mal clasificados por ambos modelos**, con el umbral de entrenamiento de 9.10.4.6.2:
  - un **falso negativo**: un préstamo que cayó en default y el modelo consideró de bajo riesgo;
  - un **falso positivo**: un préstamo que se pagó y el modelo consideró riesgoso.

  En cada caso se toma el error "más seguro" según el modelo de scikit-learn.

````python
import shutil
from lime.lime_tabular import LimeTabularExplainer

con_prob_sp = [m for m in tabla_resultados_spark.index if m != 'LinearSVC']
nombre_lime_spark = con_prob_sp[0]
modelo_lime_spark = resultados_spark[nombre_lime_spark]['modelo']
print('Modelo explicado en scikit-learn:', nombre_lime_sklearn, f"(AUC = {resultados[nombre_lime_sklearn]['ROC_AUC']:.4f})")
print('Modelo explicado en PySpark:     ', nombre_lime_spark, f"(AUC = {resultados_spark[nombre_lime_spark]['ROC_AUC']:.4f})")

pred_sk = score_sk[nombre_lime_sklearn] >= umbral_sk[nombre_lime_sklearn]
pred_sp = score_sp[nombre_lime_spark] >= umbral_sp[nombre_lime_spark]
fn_ambos = np.flatnonzero((y_comun == 1) & ~pred_sk & ~pred_sp)
fp_ambos = np.flatnonzero((y_comun == 0) & pred_sk & pred_sp)
i_fn = fn_ambos[np.argmin(score_sk[nombre_lime_sklearn][fn_ambos])]
i_fp = fp_ambos[np.argmax(score_sk[nombre_lime_sklearn][fp_ambos])]
instancias = {'Falso negativo': int(comun['id'].iloc[i_fn]), 'Falso positivo': int(comun['id'].iloc[i_fp])}
print(f'Falsos negativos comunes a ambos modelos: {len(fn_ambos):,} | falsos positivos comunes: {len(fp_ambos):,}')
print('Préstamos elegidos (id):', instancias)
````

**Explicador para PySpark.** Aquí aparecen las limitaciones de LIME en un entorno distribuido:

1. **LIME opera en memoria local.** Necesita los datos de entrenamiento como un arreglo de NumPy para calcular sus estadísticas, pero el conjunto de entrenamiento de Spark está distribuido y **no se permite traerlo completo al *driver***. Por eso se usa una **muestra aleatoria del 2 %** del entrenamiento (`sample(fraction=0.02, seed=42)`, unas 21 mil filas). Es la única forma de usar LIME sin violar la restricción, y hace que las estadísticas de perturbación sean aproximadas.
2. **Cada explicación exige llevar datos locales de vuelta a Spark.** LIME llama a la función de predicción con 5 000 filas perturbadas que están en la memoria de Python. Hay que convertirlas en un DataFrame de Spark, pasarlas por el modelo y traer las probabilidades, con la sobrecarga de serialización y planificación de Spark. En este equipo, además, `spark.createDataFrame` con datos de Python **no funciona**: requiere los procesos auxiliares de Python de Spark (*Python workers*), que en Windows con PySpark 4.0.1 no logran iniciarse. Por eso las perturbaciones se escriben en un **CSV temporal**, que Spark lee directamente desde la JVM y ensambla con `VectorAssembler`. Con muchas instancias a explicar, este costo se multiplica.

- **Coherencia entre entornos:** las variables y los signos de ambas explicaciones son **casi los mismos** (tasa, plazo, antigüedad laboral, propósito, FICO, monto). Los dos GBT aprendieron la misma relación, lo que es consistente con su AUC casi idéntico.
- **Limitaciones observadas:**
  - El **R² del modelo lineal local es solo 0.45-0.53**, así que la aproximación lineal explica apenas la mitad del comportamiento local del modelo.
  - En el falso positivo, la predicción local (0.34-0.39) **subestima mucho** la del modelo (0.82-0.87): cerca de los extremos, un modelo lineal no reproduce bien las probabilidades.
  - Algunas reglas con dummies, como `addr_state_OR=0` empujando hacia el default, son **artefactos del muestreo de perturbaciones** y no deben interpretarse como causas.
  - Las explicaciones dependen de la discretización en cuartiles y de la semilla.
- **Limitaciones en el entorno distribuido:**
  - Para Spark hubo que traer una **muestra del 2 %** del entrenamiento al *driver*, porque LIME trabaja en memoria local.
  - Hubo que pasar las 5 000 perturbaciones por un **CSV intermedio**, ya que los *Python workers* no funcionan en este equipo.
  - Cada explicación tardó **3-4 s en Spark frente a 0.1-0.2 s en scikit-learn**, unas 20 veces más. Para explicar miles de préstamos, este costo sería prohibitivo en Spark, a menos que se use una versión distribuida de las explicaciones.

  - En términos de precisión, **ningún entorno es claramente superior**: con el mismo algoritmo y los mismos hiperparámetros, ambos producen modelos equivalentes.
- **Velocidad:** en este equipo **scikit-learn fue más rápido en todo**, salvo en la lectura del CSV.

  | Etapa | scikit-learn | PySpark |
  |---|---|---|
  | Preprocesamiento | 56 s | 240 s |
  | Entrenamiento de los 6 modelos | 7 367 s (~2.0 h) | 14 026 s (~3.9 h) |
  | Predicción | 14 s | 146 s, más 14 s de transferencia al *driver* |

  - Spark fue entre 1.7 y 14 veces más lento según el modelo. La diferencia más pequeña fue en gradient boosting (6 239 s frente a 10 435 s) y la mayor en Naive Bayes (6 s frente a 87 s).
  - Ojo: los tiempos de scikit-learn se midieron con `n_jobs=4` y los de Spark con 12 núcleos, y ambos con el equipo cerca del límite de RAM.
- **Por qué PySpark no fue más rápido:**
  - **El dataset cabe en la memoria de un solo equipo.** La matriz de entrenamiento ocupa 255 MB, así que scikit-learn trabaja directamente sobre arreglos de NumPy.
  - **En modo local, Spark no tiene más recursos que scikit-learn**, pero sí más costos fijos: planificar tareas en cada iteración, serializar datos entre la JVM y Python, y usar objetos Java por fila y por nodo.
  - El `CrossValidator` entrena las combinaciones **una tras otra** (`parallelism=1`).
  - La memoria disponible obligó a usar 4 GB de heap y a paginar a disco.
- **Capacidad de cómputo insuficiente:** varias decisiones del proyecto se deben a los límites del hardware (7.7 GB de RAM, un solo equipo):
  - `n_jobs=4` en scikit-learn;
  - 4 GB de heap y `maxMemoryInMB=32` en Spark;
  - temporales de Spark en otro disco;
  - la **exclusión de RandomForest 50 × 15 y 100 × 15 en Spark**, que habrían tomado unas 6 horas adicionales.

  Con más memoria (o un clúster) el grid de Spark podría completarse, y la ventaja de Spark aparecería cuando los datos **no cupieran** en la memoria de un solo equipo. Estos puntos se retoman en la reflexión crítica del entregable.
